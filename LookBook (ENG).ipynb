### Context and Objectives
LookBook is a second-hand clothing marketplace designed to promote a more sustainable consumption model, aligned with the **UN Sustainable Development Goals 12 (Responsible Consumption)** and **13 (Climate Action)**.

The project is based on the idea that reusing garments—when supported by clear information about materials, origin, and durability—can represent a concrete alternative to fast fashion models with high environmental impact.

### Project Description
In this project, I developed an **Item Management System (IMS)** that allows users to create, organize, and analyze a collection of second-hand clothing items.

Beyond the basic features (registration, visualization, and search), the system integrates **sustainability-oriented fields and metrics**, such as:
- garment materials and fabrics  
- origin (store or brand of purchase)  
- availability for swap or donation  
- ecological score (EcoScore)  
- estimated environmental impact avoided through reuse  

The goal is not just to manage data, but to **encourage more conscious choices**, showing how second-hand consumption can reduce waste and environmental impact compared to buying new items.

### Main Features
The system allows users to:
- register new items with extended attributes (style, materials, origin)
- view and filter items with advanced options
- compare items from a “green” perspective
- generate statistics and insights on the collection
- suggest more sustainable items based on environmental criteria

The project is fully developed in **Python**, using basic data structures (lists and dictionaries), functions, conditional logic, and loops, and is designed to run inside a **Jupyter Notebook**.

```python
# In this section I define extra fields and useful options to make the marketplace more "green"
# and more complete than a simple Vinted-like clone.
# The idea is to show that LookBook encourages sustainable choices (UN 12 and 13).

#🌱 Green setup: styles, materials, eco scores

# Some examples of styles/categories to enrich the dataset (streetwear, rave, luxury fashion, etc.)
STYLE_TAGS = [
    "streetwear", "hip-hop", "rave", "clubbing", "techwear", "everyday",
    "hipster", "minimal", "vintage", "workwear", "alta moda", "outdoor",
    "skate", "y2k", "gorpcore", "business", "casual", "athleisure"
]

# Typical materials: I also include "green" options
MATERIALS = [
    "cotton", "organic cotton", "recycled cotton",
    "polyester", "recycled polyester",
    "wool", "linen", "hemp",
    "viscose", "tencel/lyocell",
    "leather", "vegan leather",
    "nylon", "recycled nylon",
    "denim"
]

# Source websites/stores (where the item was originally purchased)
# I also include fast fashion and more "premium" brands for comparison.
SOURCE_SITES = [
    "Shein", "Zara", "H&M", "Primark", "Boohoo",
    "ASOS", "Nike", "Adidas", "Supreme",
    "Gucci", "Prada", "Stone Island", "Patagonia",
    "Uniqlo", "COS", "Local boutique", "Thrift shop", "Unknown"
]

# Simple "eco rating" map for materials and sources.
# I am not trying to make perfect science here: I just need a coherent logic
# to encourage greener choices.
MATERIAL_ECO_POINTS = {
    "hemp": 5,
    "linen": 4,
    "tencel/lyocell": 4,
    "organic cotton": 4,
    "recycled cotton": 4,
    "recycled polyester": 3,
    "recycled nylon": 3,
    "cotton": 2,
    "denim": 2,
    "wool": 2,
    "polyester": 1,
    "nylon": 1,
    "viscose": 1,
    "leather": 1,
    "vegan leather": 2  # slightly above leather, but without pushing propaganda
}

SOURCE_ECO_POINTS = {
    # This is not a moral judgment: it is a way to highlight
    # fast fashion vs quality/durability dynamics.
    "Shein": 0,
    "Boohoo": 0,
    "Primark": 0,
    "H&M": 1,
    "Zara": 1,
    "ASOS": 1,
    "Uniqlo": 2,
    "COS": 2,
    "Local boutique": 3,
    "Thrift shop": 4,  # second-hand / thrift stores
    "Patagonia": 4,
    "Nike": 2,
    "Adidas": 2,
    "Supreme": 2,
    "Stone Island": 2,
    "Gucci": 2,
    "Prada": 2,
    "Unknown": 1
}

# Indicative factors for "avoided impact" when buying used instead of new
# They are intentionally simple: the goal is to provide a "green" insight in the project.
CATEGORY_IMPACT_FACTOR = {
    "t-shirt": 1.0,
    "hoodie": 2.0,
    "jeans": 2.5,
    "jacket": 3.0,
    "coat": 4.0,
    "shoes": 2.5,
    "dress": 2.5,
    "bag": 3.0,
    "accessories": 0.8
}

def eco_score(category, material, source_site, condition):
    # Here I create a simple "eco" score to encourage more sustainable choices.
    # It is not a scientific value, but a coherent index that combines
    # materials, origin, and item condition.

    base = 5  # base score because the item is second-hand

    mat_points = MATERIAL_ECO_POINTS.get(material, 1)
    src_points = SOURCE_ECO_POINTS.get(source_site, 1)

    # Condition: the closer it is to new, the more likely it is to last longer
    # (more effective reuse)
    condition_points = {
        "new": 3,
        "like new": 3,
        "good": 2,
        "worn": 1
    }.get(condition, 1)

    # Small bonus if the category is "heavier" (coat/jacket etc.),
    # because reuse matters more there.
    impact_bonus = 1 if CATEGORY_IMPACT_FACTOR.get(category, 1.0) >= 3.0 else 0

    return base + mat_points + src_points + condition_points + impact_bonus

def estimated_impact_avoided(category):
    # Simple estimate of avoided impact (arbitrary units)
    # thanks to second-hand purchasing.
    return CATEGORY_IMPACT_FACTOR.get(category, 1.0)
```

```python
def is_duplicate(items, new_item):
    """Checks whether an item is already present in the collection.

    The check is based on a combination of attributes that reasonably identify an item:
    category, brand, color, size, condition, material, source_site, style.

    Args:
        items (list[dict]): List of existing items.
        new_item (dict): Dictionary of the new item to be added (without id).

    Returns:
        bool: True if a duplicate exists, False otherwise.
    """
    keys = ["category", "brand", "color", "size", "condition", "material", "source_site", "style"]

    for item in items:
        same = True
        for k in keys:
            if str(item.get(k, "")).strip().lower() != str(new_item.get(k, "")).strip().lower():
                same = False
                break
        if same:
            return True

    return False
```

```python
#🧰
def advanced_filter():
    # In this function, I create an advanced filter with multiple criteria applied at the same time.
    # To make it more user-friendly, for some text fields I use a "contains" match
    # (instead of exact equality only), so the user does not need to type the exact same string.

    print("\nAdvanced filtering (leave blank to skip a filter)")

    category = input("Category: ").strip().lower()
    brand = input("Brand: ").strip().lower()
    color = input("Color: ").strip().lower()
    size = input("Size: ").strip().lower()
    condition = input("Condition (new/like new/good/worn): ").strip().lower()

    style = input("Style/tag (streetwear/rave/clubbing/hipster/alta moda/casual...): ").strip().lower()
    material = input("Material (cotton/organic cotton/recycled polyester/linen...): ").strip().lower()
    source_site = input("Purchased from (Shein/Zara/H&M/Thrift shop/Local boutique/Nike...): ").strip().lower()

    min_price_text = input("Minimum price (€): ").strip()
    max_price_text = input("Maximum price (€): ").strip()

    if min_price_text != "":
        min_price = float(min_price_text.replace("€", "").replace(",", "."))
    else:
        min_price = None

    if max_price_text != "":
        max_price = float(max_price_text.replace("€", "").replace(",", "."))
    else:
        max_price = None

    swap_text = input("Only items available for swap? (yes/no or blank): ").strip().lower()
    if swap_text == "yes":
        swap_filter = True
    elif swap_text == "no":
        swap_filter = False
    else:
        swap_filter = None

    donation_text = input("Only items available for donation? (yes/no or blank): ").strip().lower()
    if donation_text == "yes":
        donation_filter = True
    elif donation_text == "no":
        donation_filter = False
    else:
        donation_filter = None

    min_eco_text = input("Minimum EcoScore (number or blank): ").strip()
    min_eco = int(min_eco_text) if min_eco_text != "" else None

    results = []

    for item in items:
        # "Exact" comparisons for fields that need to be precise
        if condition and item["condition"].lower() != condition:
            continue
        if size and str(item["size"]).lower() != size:
            continue

        # "Contains" comparisons for text fields (more flexible)
        if category and category not in item["category"].lower():
            continue
        if brand and brand not in item["brand"].lower():
            continue
        if color and color not in item["color"].lower():
            continue
        if style and style not in str(item.get("style", "")).lower():
            continue
        if material and material not in str(item.get("material", "")).lower():
            continue
        if source_site and source_site not in str(item.get("source_site", "")).lower():
            continue

        if min_price is not None and item["price"] < min_price:
            continue
        if max_price is not None and item["price"] > max_price:
            continue

        if swap_filter is not None and item["swap"] != swap_filter:
            continue
        if donation_filter is not None and item.get("donation_ok", False) != donation_filter:
            continue

        if min_eco is not None and item.get("eco_score", 0) < min_eco:
            continue

        results.append(item)

    if not results:
        print("\nNo items found with these filters.")
        return

    print(f"\nResults found: {len(results)}\n")
    for item in results:
        swap_label = "Yes" if item["swap"] else "No"
        donation_label = "Yes" if item.get("donation_ok", False) else "No"
        print(
            f"ID: {item['id']} | {item['brand']} {item['category']} | "
            f"Style: {item.get('style','')} | {item['color']} | Size: {item['size']} | "
            f"{item['condition']} | €{item['price']} | "
            f"Mat: {item.get('material','')} | Src: {item.get('source_site','')} | "
            f"Swap: {swap_label} | Donation: {donation_label} | EcoScore: {item.get('eco_score','NA')} | "
            f"ImpactAvoided: {item.get('impact_avoided','NA')}"
        )

def view_items(items):
    """Displays all items in the collection in a readable format.

    Args:
        items (list[dict]): List of items with descriptive and environmental attributes.

    Returns:
        None
    """
    if not items:
        print("There are no items.")
        return
    ...
```

```python
def stats_items(items):
    """Calculates and prints statistics and insights (including green ones) about the collection.

    Args:
        items (list[dict]): List of items.

    Returns:
        None
    """
    if not items:
        print("There are no items to calculate statistics on.")
        return

    total = len(items)

    swap_count = sum(1 for item in items if item.get("swap", False))
    swap_percentage = (swap_count / total) * 100

    donation_count = sum(1 for item in items if item.get("donation_ok", False))
    donation_percentage = (donation_count / total) * 100

    prices = [item["price"] for item in items]
    min_price = min(prices)
    max_price = max(prices)
    avg_price = sum(prices) / total

    cheapest_item = min(items, key=lambda x: x["price"])
    most_expensive_item = max(items, key=lambda x: x["price"])

    by_category, by_brand, by_condition = {}, {}, {}
    by_material, by_source, by_style = {}, {}, {}

    eco_scores = []
    total_impact_avoided = 0.0

    for item in items:
        by_category[item["category"]] = by_category.get(item["category"], 0) + 1
        by_brand[item["brand"]] = by_brand.get(item["brand"], 0) + 1
        by_condition[item["condition"]] = by_condition.get(item["condition"], 0) + 1

        mat = item.get("material", "Unknown")
        src = item.get("source_site", "Unknown")
        sty = item.get("style", "Unknown")

        by_material[mat] = by_material.get(mat, 0) + 1
        by_source[src] = by_source.get(src, 0) + 1
        by_style[sty] = by_style.get(sty, 0) + 1

        eco_scores.append(item.get("eco_score", 0))
        total_impact_avoided += float(item.get("impact_avoided", 0.0))

    avg_eco = sum(eco_scores) / len(eco_scores) if eco_scores else 0

    print("\n📊 LOOKBOOK STATISTICS (UN 12 & 13)\n")
    print("Total items:", total)
    print(f"Items available for swap: {swap_count} ({swap_percentage:.1f}%)")
    print(f"Items available for donation: {donation_count} ({donation_percentage:.1f}%)")

    print(f"Minimum price: €{min_price:.2f}")
    print(f"Maximum price: €{max_price:.2f}")
    print(f"Average price: €{avg_price:.2f}")

    print("\nGreen insights:")
    print(f"- Average EcoScore: {avg_eco:.2f}")
    print(f"- Total avoided impact (estimated in relative units): {total_impact_avoided:.2f}")

    print("\nDistribution by style:")
    for k, v in by_style.items():
        print(f"- {k}: {v}")

    print("\nDistribution by category:")
    for k, v in by_category.items():
        print(f"- {k}: {v}")

    print("\nDistribution by brand:")
    for k, v in by_brand.items():
        print(f"- {k}: {v}")

    print("\nDistribution by material:")
    for k, v in by_material.items():
        print(f"- {k}: {v}")

    print("\nDistribution by source (where purchased):")
    for k, v in by_source.items():
        print(f"- {k}: {v}")

    print("\nDistribution by condition:")
    for k, v in by_condition.items():
        print(f"- {k}: {v}")

    print("\nCheapest item:")
    print_item(cheapest_item)

    print("\nMost expensive item:")
    print_item(most_expensive_item)
```

```python
#🌿
def show_green_picks():
    # In this function, I display a selection of "recommended" items using eco_score as the metric:
    # the goal is to encourage greener choices.
    # It is not perfect science, but a coherent criterion for the marketplace.

    if not items:
        print("No items available.")
        return

    top_n_text = input("How many items would you like to see? (default 5): ").strip()
    top_n = int(top_n_text) if top_n_text != "" else 5

    sorted_items = sorted(items, key=lambda x: x.get("eco_score", 0), reverse=True)
    picks = sorted_items[:top_n]

    print(f"\n🌱 Green Picks (Top {len(picks)} by EcoScore)\n")
    for item in picks:
        print(
            f"ID {item['id']} | {item['brand']} {item['category']} | "
            f"Mat: {item.get('material','')} | Src: {item.get('source_site','')} | "
            f"Cond: {item['condition']} | €{item['price']} | EcoScore: {item.get('eco_score','NA')} | "
            f"ImpactAvoided: {item.get('impact_avoided','NA')}"
        )
```

```python
#🥬
def explain_green_choice():
    # In this function, I explain why an item is more or less "green",
    # so LookBook is not just a marketplace, but also encourages more conscious consumption (UN 12/13).

    if not items:
        print("No items available.")
        return

    id_text = input("Enter the item ID: ").strip()
    if id_text == "":
        print("Invalid ID.")
        return

    target_id = int(id_text)

    target = None
    for item in items:
        if item["id"] == target_id:
            target = item
            break

    if not target:
        print("Item not found.")
        return

    material = target.get("material", "Unknown")
    source_site = target.get("source_site", "Unknown")
    score = target.get("eco_score", "NA")
    impact = target.get("impact_avoided", "NA")

    print("\n🧠 Green Explanation\n")
    print(f"Item: {target['brand']} {target['category']} | Condition: {target['condition']} | Price: €{target['price']}")
    print(f"Material: {material} | Source: {source_site}")
    print(f"EcoScore: {score} | ImpactAvoided (estimated): {impact}")

    mat_points = MATERIAL_ECO_POINTS.get(material, 1)
    src_points = SOURCE_ECO_POINTS.get(source_site, 1)

    print("\nWhy this score:")
    print(f"- Material (points): {mat_points}")
    print(f"- Source (points): {src_points}")
    print("- Second-hand: positive base value (reuse)")
    print("- Condition: better condition = longer lifespan = more effective reuse")

    print("\nSuggestion:")
    if mat_points >= 4:
        print("- Excellent material for more sustainable choices.")
    elif mat_points <= 1:
        print("- Less sustainable material: if you buy it, aim for long-term use and take good care of it so it lasts.")
    if src_points <= 1:
        print("- Typical fast fashion source: reusing it is still better than buying it new, but durability and quality should be considered.")
```

## Guide to Using the LookBook System (IMS)

The LookBook system is designed to be used through an **interactive menu** within a Jupyter Notebook.  
Interaction happens by entering numeric or text values requested by the system.

### Starting the System
To use the system correctly, you need to:
- run all definition cells (setup and functions)
- load a demo dataset
- start the main menu

Loading demo data allows you to work with a large and diverse collection of items without having to manually input them every time.

---

### Menu Structure
The main menu allows access to system features through numeric options.

**1) Load demo data**  
Populates the system with a collection of example items belonging to different styles  
(streetwear, rave, everyday, outdoor, luxury fashion) and including sustainability-related attributes  
such as materials, source, and EcoScore.

**2) Add a new item**  
Allows manual insertion of an item through guided input.  
In addition to basic attributes (category, brand, size, price), the system requires:
- item style
- main material
- origin (where it was purchased)
- availability for swap or donation  

These inputs are used to calculate an **ecological score (EcoScore)**.

**3) View all items**  
Displays the full list of items in the system.  
Each item is shown including:
- descriptive information
- EcoScore
- estimated environmental impact avoided through reuse

---

### Search and Filtering

**4) Simple search**  
Allows users to search items based on a single attribute (e.g. brand, category, or material).  
Designed for quick and direct use.

**5) Advanced filtering**  
Allows combining multiple criteria at once.  
Available filters include:
- descriptive attributes (category, brand, color, size, style)
- economic criteria (minimum and maximum price)
- functional criteria (swap and donation)
- environmental criteria (minimum EcoScore)

Text fields use a **“contains” match**, making the system more flexible  
and closer to real-world usage.  
Leaving a field empty means that filter is not applied.

---

### Analysis and Sustainability

**6) Statistics**  
Generates a summary report of the collection, including:
- total number of items
- distributions by category, brand, style, material, and source
- percentage of items available for swap or donation
- average EcoScore
- total estimated environmental impact avoided

This section connects the system to **Responsible Consumption (UN 12)**  
and **Environmental Impact Reduction (UN 13)**.

**7) Green Picks**  
Displays a selection of items with the highest EcoScore, automatically suggesting  
the most sustainable options.

---

### Exiting the System
**0) Exit**  
Ends the menu execution and closes the system.

---

- definition cells (setup and functions)

This improves code readability and ensures a clean, reproducible demo.
