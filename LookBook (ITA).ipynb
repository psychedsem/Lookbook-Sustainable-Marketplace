### Contesto e obiettivi
LookBook è un marketplace di abbigliamento second-hand pensato per promuovere un modello di consumo più sostenibile, in linea con gli **Obiettivi di Sviluppo Sostenibile ONU 12 (Consumo responsabile)** e **ONU 13 (Lotta al cambiamento climatico)**.

Il progetto nasce dall’idea che il riuso dei capi, se accompagnato da informazioni chiare su materiali, provenienza e durata, possa rappresentare un’alternativa concreta ai modelli di fast fashion ad alto impatto ambientale.

### Descrizione del progetto
In questo progetto realizzo un **Sistema di Gestione degli Elementi (SGE)** che permette di creare, organizzare e analizzare una collezione di capi di abbigliamento usato.

Oltre alle funzionalità di base (registrazione, visualizzazione e ricerca), il sistema integra **campi e metriche orientate alla sostenibilità**, come:
- materiali e tessuti del capo  
- provenienza (negozio o brand di acquisto)  
- disponibilità allo swap o alla donazione  
- punteggio ecologico (EcoScore)  
- stima dell’impatto ambientale evitato grazie al riuso  

L’obiettivo non è solo gestire i dati, ma **incentivare scelte più consapevoli**, mostrando come il second-hand possa ridurre sprechi e impatto ambientale rispetto all’acquisto di capi nuovi.

### Funzionalità principali
Il sistema consente di:
- registrare nuovi capi con attributi estesi (stile, materiali, provenienza)
- visualizzare e filtrare i capi in modo avanzato
- confrontare i capi anche dal punto di vista “green”
- generare statistiche e insight sulla collezione
- suggerire capi più sostenibili tramite criteri ambientali

Il progetto è sviluppato interamente in **Python**, utilizzando strutture dati di base (liste e dizionari), funzioni, logiche condizionali e cicli, ed è pensato per essere eseguito all’interno di un **Jupyter Notebook**.

```python
# In questa sezione definisco campi extra e opzioni utili per rendere il marketplace più "green" e più completo di un semplice clone tipo Vinted.
# L'idea è far vedere che LookBook incentiva scelte sostenibili (ONU 12 e 13).

#🌱 Setup green: stili, materiali, punteggi ecologici

# Alcuni esempi di stili/categorie per arricchire il dataset (streetwear, rave, alta moda, ecc.)
STYLE_TAGS = [
    "streetwear", "hip-hop", "rave", "clubbing", "techwear", "everyday",
    "hipster", "minimal", "vintage", "workwear", "alta moda", "outdoor",
    "skate", "y2k", "gorpcore", "business", "casual", "athleisure"
]

# Materiali tipici: includo anche opzioni "green"
MATERIALS = [
    "cotton", "organic cotton", "recycled cotton",
    "polyester", "recycled polyester",
    "wool", "linen", "hemp",
    "viscose", "tencel/lyocell",
    "leather", "vegan leather",
    "nylon", "recycled nylon",
    "denim"
]

# Siti/negozi di provenienza (dove è stato comprato il capo)
# Inserisco anche fast fashion e brand più "premium" per confronto.
SOURCE_SITES = [
    "Shein", "Zara", "H&M", "Primark", "Boohoo",
    "ASOS", "Nike", "Adidas", "Supreme",
    "Gucci", "Prada", "Stone Island", "Patagonia",
    "Uniqlo", "COS", "Local boutique", "Thrift shop", "Unknown"
]

# Mappa semplice di "eco rating" per materiali e provenienza.
# Non sto facendo scienza perfetta: mi serve una logica coerente per incentivare scelte più green.
MATERIAL_ECO_POINTS = {
    "hemp": 5,
    "linen": 4,
    "tencel/lyocell": 4,
    "organic cotton": 4,
    "recycled cotton": 4,
    "recycled polyester": 3,
    "recycled nylon": 3,
    "cotton": 2,
    "denim": 2,
    "wool": 2,
    "polyester": 1,
    "nylon": 1,
    "viscose": 1,
    "leather": 1,
    "vegan leather": 2  # metto leggermente sopra pelle, ma senza fare propaganda
}

SOURCE_ECO_POINTS = {
    # Non è un giudizio morale: è un modo per evidenziare dinamiche fast fashion vs qualità/durata.
    "Shein": 0,
    "Boohoo": 0,
    "Primark": 0,
    "H&M": 1,
    "Zara": 1,
    "ASOS": 1,
    "Uniqlo": 2,
    "COS": 2,
    "Local boutique": 3,
    "Thrift shop": 4,  # second hand / mercatini
    "Patagonia": 4,
    "Nike": 2,
    "Adidas": 2,
    "Supreme": 2,
    "Stone Island": 2,
    "Gucci": 2,
    "Prada": 2,
    "Unknown": 1
}

# Fattori indicativi di "impatto evitato" se si compra usato invece che nuovo
# Sono volutamente semplici: servono per dare un insight "green" nel progetto.
CATEGORY_IMPACT_FACTOR = {
    "t-shirt": 1.0,
    "hoodie": 2.0,
    "jeans": 2.5,
    "jacket": 3.0,
    "coat": 4.0,
    "shoes": 2.5,
    "dress": 2.5,
    "bag": 3.0,
    "accessories": 0.8
}

def eco_score(category, material, source_site, condition):
    # Qui creo un punteggio "eco" semplice per incentivare scelte più sostenibili.
    # Non è un valore scientifico, ma un indice coerente che combina materiali, provenienza e stato del capo.

    base = 5  # base per il fatto che è second-hand

    mat_points = MATERIAL_ECO_POINTS.get(material, 1)
    src_points = SOURCE_ECO_POINTS.get(source_site, 1)

    # Condizione: più è vicino al nuovo, più è probabile che duri a lungo (riuso più efficace)
    condition_points = {
        "new": 3,
        "like new": 3,
        "good": 2,
        "worn": 1
    }.get(condition, 1)

    # Bonus leggero se la categoria è “pesante” (coat/jacket ecc), perché il riuso lì pesa di più.
    impact_bonus = 1 if CATEGORY_IMPACT_FACTOR.get(category, 1.0) >= 3.0 else 0

    return base + mat_points + src_points + condition_points + impact_bonus

def estimated_impact_avoided(category):
    # Stima semplice di impatto evitato (unità arbitrarie) grazie all'acquisto second-hand.
    return CATEGORY_IMPACT_FACTOR.get(category, 1.0)
```

```python
def is_duplicate(items, new_item):
    """Controlla se un capo è già presente nella collezione.

    Il controllo è basato su una combinazione di attributi che identificano un capo in modo ragionevole:
    category, brand, color, size, condition, material, source_site, style.

    Args:
        items (list[dict]): Lista dei capi già presenti.
        new_item (dict): Dizionario del nuovo capo da inserire (senza id).

    Returns:
        bool: True se esiste un duplicato, False altrimenti.
    """
    keys = ["category", "brand", "color", "size", "condition", "material", "source_site", "style"]

    for item in items:
        same = True
        for k in keys:
            if str(item.get(k, "")).strip().lower() != str(new_item.get(k, "")).strip().lower():
                same = False
                break
        if same:
            return True

    return False
```

```python
#🧰
def advanced_filter():
    # In questa funzione creo un filtraggio avanzato con più criteri simultanei.
    # Per renderlo più user-friendly, per alcuni campi testuali uso il match "contiene"
    # (e non solo uguaglianza perfetta), così l'utente non deve scrivere la stringa identica.

    print("\nFiltraggio avanzato (lascia vuoto per saltare un filtro)")

    category = input("Categoria: ").strip().lower()
    brand = input("Brand: ").strip().lower()
    color = input("Colore: ").strip().lower()
    size = input("Taglia: ").strip().lower()
    condition = input("Condizione (new/like new/good/worn): ").strip().lower()

    style = input("Stile/tag (streetwear/rave/clubbing/hipster/alta moda/casual...): ").strip().lower()
    material = input("Materiale (cotton/organic cotton/recycled polyester/linen...): ").strip().lower()
    source_site = input("Comprato su (Shein/Zara/H&M/Thrift shop/Local boutique/Nike...): ").strip().lower()

    min_price_text = input("Prezzo minimo (€): ").strip()
    max_price_text = input("Prezzo massimo (€): ").strip()

    if min_price_text != "":
        min_price = float(min_price_text.replace("€", "").replace(",", "."))
    else:
        min_price = None

    if max_price_text != "":
        max_price = float(max_price_text.replace("€", "").replace(",", "."))
    else:
        max_price = None

    swap_text = input("Solo capi disponibili per swap? (yes/no o vuoto): ").strip().lower()
    if swap_text == "yes":
        swap_filter = True
    elif swap_text == "no":
        swap_filter = False
    else:
        swap_filter = None

    donation_text = input("Solo capi disponibili per donazione? (yes/no o vuoto): ").strip().lower()
    if donation_text == "yes":
        donation_filter = True
    elif donation_text == "no":
        donation_filter = False
    else:
        donation_filter = None

    min_eco_text = input("EcoScore minimo (numero o vuoto): ").strip()
    min_eco = int(min_eco_text) if min_eco_text != "" else None

    results = []

    for item in items:
        # Confronti "esatti" per campi che devono essere precisi
        if condition and item["condition"].lower() != condition:
            continue
        if size and str(item["size"]).lower() != size:
            continue

        # Confronti "contiene" per campi testuali (più tollerante)
        if category and category not in item["category"].lower():
            continue
        if brand and brand not in item["brand"].lower():
            continue
        if color and color not in item["color"].lower():
            continue
        if style and style not in str(item.get("style", "")).lower():
            continue
        if material and material not in str(item.get("material", "")).lower():
            continue
        if source_site and source_site not in str(item.get("source_site", "")).lower():
            continue

        if min_price is not None and item["price"] < min_price:
            continue
        if max_price is not None and item["price"] > max_price:
            continue

        if swap_filter is not None and item["swap"] != swap_filter:
            continue
        if donation_filter is not None and item.get("donation_ok", False) != donation_filter:
            continue

        if min_eco is not None and item.get("eco_score", 0) < min_eco:
            continue

        results.append(item)

    if not results:
        print("\nNessun capo trovato con questi filtri.")
        return

    print(f"\nRisultati trovati: {len(results)}\n")
    for item in results:
        swap_label = "Sì" if item["swap"] else "No"
        donation_label = "Sì" if item.get("donation_ok", False) else "No"
        print(
            f"ID: {item['id']} | {item['brand']} {item['category']} | "
            f"Stile: {item.get('style','')} | {item['color']} | Taglia: {item['size']} | "
            f"{item['condition']} | €{item['price']} | "
            f"Mat: {item.get('material','')} | Src: {item.get('source_site','')} | "
            f"Swap: {swap_label} | Donazione: {donation_label} | EcoScore: {item.get('eco_score','NA')} | "
            f"ImpactAvoided: {item.get('impact_avoided','NA')}"
        )

def view_items(items):
    """Mostra in modo leggibile tutti i capi presenti nella collezione.

    Args:
        items (list[dict]): Lista di capi con attributi descrittivi e ambientali.

    Returns:
        None
    """
    if not items:
        print("Non ci sono capi.")
        return
    ...
```

```python
def stats_items(items):
    """Calcola e stampa statistiche e insight (anche green) sulla collezione.

    Args:
        items (list[dict]): Lista dei capi.

    Returns:
        None
    """
    if not items:
        print("Non ci sono capi su cui calcolare statistiche.")
        return

    total = len(items)

    swap_count = sum(1 for item in items if item.get("swap", False))
    swap_percentage = (swap_count / total) * 100

    donation_count = sum(1 for item in items if item.get("donation_ok", False))
    donation_percentage = (donation_count / total) * 100

    prices = [item["price"] for item in items]
    min_price = min(prices)
    max_price = max(prices)
    avg_price = sum(prices) / total

    cheapest_item = min(items, key=lambda x: x["price"])
    most_expensive_item = max(items, key=lambda x: x["price"])

    by_category, by_brand, by_condition = {}, {}, {}
    by_material, by_source, by_style = {}, {}, {}

    eco_scores = []
    total_impact_avoided = 0.0

    for item in items:
        by_category[item["category"]] = by_category.get(item["category"], 0) + 1
        by_brand[item["brand"]] = by_brand.get(item["brand"], 0) + 1
        by_condition[item["condition"]] = by_condition.get(item["condition"], 0) + 1

        mat = item.get("material", "Unknown")
        src = item.get("source_site", "Unknown")
        sty = item.get("style", "Unknown")

        by_material[mat] = by_material.get(mat, 0) + 1
        by_source[src] = by_source.get(src, 0) + 1
        by_style[sty] = by_style.get(sty, 0) + 1

        eco_scores.append(item.get("eco_score", 0))
        total_impact_avoided += float(item.get("impact_avoided", 0.0))

    avg_eco = sum(eco_scores) / len(eco_scores) if eco_scores else 0

    print("\n📊 STATISTICHE LOOKBOOK (ONU 12 & 13)\n")
    print("Totale capi:", total)
    print(f"Capi disponibili per swap: {swap_count} ({swap_percentage:.1f}%)")
    print(f"Capi disponibili per donazione: {donation_count} ({donation_percentage:.1f}%)")

    print(f"Prezzo minimo: €{min_price:.2f}")
    print(f"Prezzo massimo: €{max_price:.2f}")
    print(f"Prezzo medio:  €{avg_price:.2f}")

    print("\nInsight green:")
    print(f"- EcoScore medio: {avg_eco:.2f}")
    print(f"- Impatto evitato totale (stima in unità relative): {total_impact_avoided:.2f}")

    print("\nDistribuzione per stile:")
    for k, v in by_style.items():
        print(f"- {k}: {v}")

    print("\nDistribuzione per categoria:")
    for k, v in by_category.items():
        print(f"- {k}: {v}")

    print("\nDistribuzione per brand:")
    for k, v in by_brand.items():
        print(f"- {k}: {v}")

    print("\nDistribuzione per materiale:")
    for k, v in by_material.items():
        print(f"- {k}: {v}")

    print("\nDistribuzione per provenienza (dove comprato):")
    for k, v in by_source.items():
        print(f"- {k}: {v}")

    print("\nDistribuzione per condizione:")
    for k, v in by_condition.items():
        print(f"- {k}: {v}")

    print("\nCapo più economico:")
    print_item(cheapest_item)

    print("\nCapo più caro:")
    print_item(most_expensive_item)
```

```python
#🌿
def show_green_picks():
    # In questa funzione mostro una selezione di capi "consigliati" usando l'eco_score come metrica: serve per incentivare scelte più green.
    # Non è scienza perfetta: è un criterio coerente per il marketplace.

    if not items:
        print("Non ci sono capi disponibili.")
        return

    top_n_text = input("Quanti capi vuoi vedere? (default 5): ").strip()
    top_n = int(top_n_text) if top_n_text != "" else 5

    sorted_items = sorted(items, key=lambda x: x.get("eco_score", 0), reverse=True)
    picks = sorted_items[:top_n]

    print(f"\n🌱 Green Picks (Top {len(picks)} per EcoScore)\n")
    for item in picks:
        print(
            f"ID {item['id']} | {item['brand']} {item['category']} | "
            f"Mat: {item.get('material','')} | Src: {item.get('source_site','')} | "
            f"Cond: {item['condition']} | €{item['price']} | EcoScore: {item.get('eco_score','NA')} | "
            f"ImpactAvoided: {item.get('impact_avoided','NA')}"
        )
```

```python
#🥬
def explain_green_choice():
    # In questa funzione spiego perché un capo è più o meno "green", 
    # così LookBook non è solo un marketplace, ma educa a consumi più consapevoli (ONU 12/13).

    if not items:
        print("Non ci sono capi disponibili.")
        return

    id_text = input("Inserisci l'ID del capo: ").strip()
    if id_text == "":
        print("ID non valido.")
        return

    target_id = int(id_text)

    target = None
    for item in items:
        if item["id"] == target_id:
            target = item
            break

    if not target:
        print("Capo non trovato.")
        return

    material = target.get("material", "Unknown")
    source_site = target.get("source_site", "Unknown")
    score = target.get("eco_score", "NA")
    impact = target.get("impact_avoided", "NA")

    print("\n🧠 Green Explanation\n")
    print(f"Capo: {target['brand']} {target['category']} | Condizione: {target['condition']} | Prezzo: €{target['price']}")
    print(f"Materiale: {material} | Provenienza: {source_site}")
    print(f"EcoScore: {score} | ImpactAvoided (stima): {impact}")

    mat_points = MATERIAL_ECO_POINTS.get(material, 1)
    src_points = SOURCE_ECO_POINTS.get(source_site, 1)

    print("\nPerché questo punteggio:")
    print(f"- Materiale (punti): {mat_points}")
    print(f"- Provenienza (punti): {src_points}")
    print("- Second-hand: base positiva (riuso)")
    print("- Condizione: migliore condizione = più durata = più riuso efficace")

    print("\nSuggerimento:")
    if mat_points >= 4:
        print("- Ottimo materiale per scelte più sostenibili.")
    elif mat_points <= 1:
        print("- Materiale meno sostenibile: se lo prendi, puntaci per uso lungo e cura per farlo durare.")
    if src_points <= 1:
        print("- Provenienza tipica fast fashion: riusarlo è comunque meglio che comprarlo nuovo, ma valuta durata/qualità.")
```

## Guida all’utilizzo del sistema LookBook (SGE)

Il sistema LookBook è progettato per essere utilizzato tramite un **menu interattivo** all’interno del Jupyter Notebook.
L’interazione avviene inserendo valori numerici o testuali richiesti dal sistema.

### Avvio del sistema
Per utilizzare correttamente il sistema è necessario:
- eseguire tutte le celle di definizione (setup e funzioni)
- caricare un dataset di esempio
- avviare il menu principale

Il caricamento dei dati di esempio permette di lavorare su una collezione ampia e variegata di capi senza doverli inserire manualmente ogni volta.

---

### Struttura del menu
Il menu principale consente di accedere alle funzionalità del sistema tramite opzioni numeriche.

**1) Carica dati demo**  
Popola il sistema con una collezione di capi di esempio appartenenti a stili diversi
(streetwear, rave, everyday, outdoor, alta moda) e con attributi legati alla sostenibilità
come materiali, provenienza ed EcoScore.

**2) Aggiungi un nuovo capo**  
Permette l’inserimento manuale di un capo tramite input guidato.  
Oltre agli attributi base (categoria, brand, taglia, prezzo), il sistema richiede:
- stile del capo
- materiale principale
- provenienza (dove è stato acquistato)
- disponibilità allo swap o alla donazione  

Queste informazioni vengono utilizzate per calcolare un **punteggio ecologico (EcoScore)**.

**3) Visualizza tutti i capi**  
Mostra l’elenco completo dei capi presenti nel sistema.  
Ogni capo viene visualizzato includendo:
- informazioni descrittive
- EcoScore
- stima dell’impatto ambientale evitato grazie al riuso

---

### Ricerca e filtraggio

**4) Ricerca semplice**  
Consente di cercare i capi in base a un singolo attributo (ad esempio brand, categoria o materiale).
È pensata per un utilizzo rapido e diretto.

**5) Filtraggio avanzato**  
Permette di combinare più criteri contemporaneamente.  
I filtri disponibili includono:
- attributi descrittivi (categoria, brand, colore, taglia, stile)
- criteri economici (prezzo minimo e massimo)
- criteri funzionali (swap e donazione)
- criteri ambientali (EcoScore minimo)

I campi testuali utilizzano un confronto di tipo **“contiene”**, in modo da rendere il sistema
più tollerante agli input dell’utente e più vicino a un utilizzo reale.  
Lasciando vuoto un campo, il relativo filtro non viene applicato.

---

### Analisi e sostenibilità

**6) Statistiche**  
Genera un report riepilogativo sulla collezione, includendo:
- numero totale di capi
- distribuzioni per categoria, brand, stile, materiale e provenienza
- percentuale di capi disponibili per swap o donazione
- EcoScore medio
- stima complessiva dell’impatto ambientale evitato

Questa sezione collega il sistema agli obiettivi di **consumo responsabile (ONU 12)**
e **riduzione dell’impatto ambientale (ONU 13)**.

**7) Green Picks**  
Mostra una selezione dei capi con EcoScore più elevato, suggerendo automaticamente
le opzioni potenzialmente più sostenibili.

**0) Esci**  
Termina l’esecuzione del menu e conclude l’utilizzo del sistema.
